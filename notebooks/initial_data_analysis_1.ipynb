```python
# Import کتابخانه‌ها
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer

# تنظیمات نمایش
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
pd.set_option('display.float_format', '{:.2f}'.format)
```

## ۱. بارگذاری داده

دیتاست Breast Cancer Wisconsin از scikit-learn بارگذاری می‌شود.

```python
# بارگذاری دیتاست
data = load_breast_cancer()

# ساخت DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# اضافه کردن Target
df['target'] = data.target

# نمایش ۵ ردیف اول
df.head()
```

## ۲. بررسی ساختار داده

```python
# ابعاد داده
print("Shape:", df.shape)
print()
print("تعداد نمونه‌ها:", df.shape[0])
print("تعداد ستون‌ها:", df.shape[1])
print("تعداد ویژگی‌ها (بدون target):", df.shape[1] - 1)
```

یافته ۱: داده شامل ۵۶۹ نمونه و ۳۱ ستون است (۳۰ ویژگی + ۱ ستون هدف).

```python
# نام ستون‌ها
print("نام ستون‌ها:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")
```

یافته ۲: ستون‌ها شامل ۳۰ ویژگی عددی (با نام‌هایی مثل mean radius، mean texture) و یک ستون target هستند.

## ۳. بررسی نوع داده‌ها

```python
# نوع داده‌های هر ستون
print("توزیع نوع داده‌ها:")
print(df.dtypes.value_counts())
print()
print("نوع تمام ستون‌ها:")
df.dtypes
```

یافته ۳: ۳۰ ویژگی از نوع float64 (عدد اعشاری) و ستون target از نوع int64 (عدد صحیح) است.

```python
# اطلاعات کامل
df.info()
```

## ۴. بررسی متغیر هدف (Target)

```python
# توزیع کلاس‌های Target
print("توزیع کلاس‌های Target:")
print(df['target'].value_counts())
print()
print("درصد:")
print(df['target'].value_counts(normalize=True) * 100)
print()
print("نام کلاس‌ها:")
for i, name in enumerate(data.target_names):
    print(f"{i} = {name}")
```

## ۵. بررسی مقادیر گمشده

```python
# بررسی مقادیر گمشده
missing = df.isna().sum()

print("تعداد مقادیر گمشده در هر ستون:")
print(missing)
print()
print("مجموع مقادیر گمشده:", missing.sum())
```

یافته ۵: داده هیچ مقدار گمشده‌ای ندارد. مجموع مقادیر گمشده = ۰.

## ۶. بررسی ویژگی‌های ثابت یا مشکوک

```python
# بررسی ویژگی‌های ثابت (واریانس صفر)
constant_features = []

for col in df.columns:
    if df[col].nunique() == 1:
        constant_features.append(col)

print("ویژگی‌های ثابت (تنها یک مقدار یکتا):")
print(constant_features if constant_features else "هیچ ویژگی ثابتی وجود ندارد")
```

یافته ۶: هیچ ویژگی ثابتی (با واریانس صفر) در داده وجود ندارد.

```python
# بررسی ستون‌هایی که ممکنه مشکوک باشن
print("ستون‌هایی که ممکنه شناسه باشن (nunique = تعداد ردیف):")
id_cols = []
for col in df.columns:
    if df[col].nunique() == df.shape[0]:
        id_cols.append(col)

print(id_cols if id_cols else "هیچ ستون شناسه‌ای وجود ندارد")
print()

# آمار کلی هر ستون
print("خلاصهٔ آماری همهٔ ستون‌ها:")
df.describe()
```

یافته ۷: هیچ ستون شناسه‌ای (id) در داده وجود ندارد. همهٔ ستون‌ها ویژگی عددی معتبری هستند (به‌جز target).

## ۷. ستون‌هایی که نباید به مدل داده شوند

ستون target نباید به عنوان ورودی مدل داده شود، چون:
- این همون چیزیه که می‌خوایم پیش‌بینی کنیم.
- اگه به مدل بدیم، مدل فقط از روی جواب، جواب می‌ده (Data Leakage).

پس X = تمام ستون‌ها به‌جز target، و y = ستون target.

```python
# جدا کردن X و y
X = df.drop('target', axis=1)   # ۳۰ ویژگی
y = df['target']                 # هدف

print("Shape X:", X.shape)
print("Shape y:", y.shape)
```

یافته ۸: X شامل ۳۰ ویژگی است (569, 30) و y شامل متغیر هدف (569,) است. هیچ ستون اضافی‌ای وجود ندارد.
