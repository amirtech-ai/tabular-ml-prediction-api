## ۱. بارگذاری داده

دیتاست Breast Cancer Wisconsin از scikit-learn بارگذاری می‌شود.

```python
# بارگذاری دیتاست
data = load_breast_cancer()

# ساخت DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# اضافه کردن Target
df['target'] = data.target
df = df.rename(columns={'target': 'target'})
# نمایش ۵ ردیف اول
df.head()
```

## ۵. بررسی مقادیر گمشده

```python
# بررسی مقادیر گمشده
missing = df.isna().sum()

print("تعداد مقادیر گمشده در هر ستون:")
print(missing)
print()
print("مجموع مقادیر گمشده:", missing.sum())
```

## ۶. بررسی ویژگی‌های ثابت یا مشکوک

```python
# بررسی ویژگی‌های ثابت (واریانس صفر)
constant_features = []

for col in df.columns:
    if df[col].nunique() == 1:
        constant_features.append(col)

print("ویژگی‌های ثابت (تنها یک مقدار یکتا):")
print(constant_features if constant_features else "هیچ ویژگی ثابتی وجود ندارد")
```

یافته ۶: هیچ ویژگی ثابتی (با واریانس صفر) در داده وجود ندارد.

```python
# بررسی ستون‌هایی که ممکنه مشکوک باشن
print("ستون‌هایی که ممکنه شناسه باشن (nunique = تعداد ردیف):")
id_cols = []
for col in df.columns:
    if df[col].nunique() == df.shape[0]:
        id_cols.append(col)

print(id_cols if id_cols else "هیچ ستون شناسه‌ای وجود ندارد")
print()

# آمار کلی هر ستون
print("خلاصهٔ آماری همهٔ ستون‌ها:")
df.describe()
```

یافته ۷: هیچ ستون شناسه‌ای (id) در داده وجود ندارد. همهٔ ستون‌ها ویژگی عددی معتبری هستند (به‌جز target).

```python
# تکرار import ها برای تعریف روش جدید بارگذار dataframe
import pandas as pd
from sklearn.datasets import load_breast_cancer

# بارگذاری دیتاست
dataset = load_breast_cancer(as_frame=True)
df2 = dataset.frame.copy()

# جداسازی ویژگی‌ها و هدف
X = df2.drop(columns=["target"])
y = df2["target"]

# بررسی ابعاد و کلاس‌ها
print("Features:", X.shape)
print("Target:", y.shape)
print("Classes:", y.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split

# تقسیم داده: ۸۰٪ آموزش، ۲۰٪ آزمون
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
X_train_3, X_test_3, y_train_3, y_test_3 = train_test_split(
    X, y, test_size=0.20, random_state=7, stratify=y
)

print("Indexes identical with random_state=7?", X_train.index.equals(X_train_3.index))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# ساخت Pipeline با یک مرحله: StandardScaler
preprocessing_pipeline = Pipeline([
    ("scaler", StandardScaler())
])

# روی داده آموزش: fit_transform
X_train_scaled = preprocessing_pipeline.fit_transform(X_train)

# روی داده آزمون: فقط transform
X_test_scaled = preprocessing_pipeline.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)
```
