نمایش درصدها با علامت درصد

در این بخش، درصد فراوانی هر کلاس را محاسبه کرده و با استفاده از astype(str) به متن تبدیل کردیم تا بتوانیم علامت % را به آن بچسبانیم. این کار برای گزارش‌نویسی و نمایش زیباتر نتایج مفید است.

```python
# ۱. محاسبه درصد هر کلاس (بدون تبدیل به متن)
proportions = df['target'].value_counts(normalize=True) * 100

# ۲. حلقه روی هر کلاس
for target_class, percent in proportions.items():
    # تشخیص نام کلاس
    if target_class == 0:
        label = "بیمار است (malignant)"
    elif target_class == 1:
        label = "سالم است (benign)"
    else:
        label = "نامشخص"
    
    # چاپ خروجی با قالب‌بندی زیبا
    print(f"کلاس {target_class}: {percent:.2f}% ← {label}")
```

بررسی بازتولیدپذیری تقسیم داده

با اجرای مجدد train_test_split با random_state=42، مشاهده می‌کنیم که ایندکس‌های مجموعه آموزش دقیقاً یکسان هستند (True). این یعنی تقسیم داده قابل بازتولید است. اما با تغییر random_state به 7، ایندکس‌ها تغییر می‌کنند (False). این آزمایش نشان می‌دهد که random_state چگونه کار می‌کند و چرا برای پروژه‌های علمی حیاتی است.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# ساخت Pipeline با یک مرحله: StandardScaler
preprocessing_pipeline = Pipeline([
    ("scaler", StandardScaler())
])

# روی داده آموزش: fit_transform
X_train_scaled = preprocessing_pipeline.fit_transform(X_train)

# روی داده آزمون: فقط transform
X_test_scaled = preprocessing_pipeline.transform(X_test)

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled test shape:", X_test_scaled.shape)
```

ساخت Pipeline استانداردسازی

در این بخش، یک Pipeline اولیه با StandardScaler ساختیم. این Pipeline را روی X_train با fit_transform برازش کردیم و روی X_test فقط با transform اعمال کردیم. این ترتیب عملیات از Data Leakage جلوگیری می‌کند، چون پارامترهای استانداردسازی (میانگین و انحراف معیار) فقط از داده آموزش یاد گرفته می‌شوند.

بررسی نهایی ابعاد و هم‌ترازی

در این بخش، ابعاد داده‌ها را قبل و بعد از استانداردسازی بررسی کردیم. تعداد ردیف‌های X_train با y_train و X_test با y_test برابر است. همچنین استانداردسازی تغییری در ابعاد داده ایجاد نکرده است. این موارد نشان می‌دهد که مراحل آماده‌سازی داده به درستی انجام شده‌اند.

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

# ساخت Baseline: همیشه کلاس پرتکرار را پیش‌بینی کن
baseline = DummyClassifier(strategy="most_frequent")

# آموزش Baseline روی داده آموزش
baseline.fit(X_train, y_train)

# پیش‌بینی روی داده آزمون
baseline_predictions = baseline.predict(X_test)

# محاسبه دقت
baseline_accuracy = accuracy_score(y_test, baseline_predictions)

print(baseline_predictions)
print("Baseline Accuracy:", baseline_accuracy)
```

ساخت Baseline

Baseline یک نقطه‌ی مقایسه‌ی ساده است. در اینجا از DummyClassifier(strategy="most_frequent") استفاده کردیم که همیشه کلاس پرتکرار در داده آموزش (کلاس 1 = benign) را پیش‌بینی می‌کند. دقت آن روی داده آزمون ۶۳.۱۵٪ است. هر مدل واقعی باید از این عدد بهتر عمل کند تا ارزشمند باشد.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# ساخت Pipeline: استانداردساز + مدل
model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000))
])

# آموزش مدل روی داده آموزش
model.fit(X_train, y_train)

print("Model trained successfully")
```

ساخت و آموزش Pipeline مدل

در این بخش، Pipeline نهایی را با دو مرحله ساختیم: StandardScaler برای استانداردسازی و LogisticRegression برای طبقه‌بندی. با اجرای model.fit(X_train, y_train)، هر دو مرحله به ترتیب روی داده آموزش برازش می‌شوند. این ساختار منظم و قابل تکرار است و از Data Leakage جلوگیری می‌کند.

```python
# تولید پیش‌بینی
predictions = model.predict(X_test)

print("First 10 predictions:")
print(predictions[:10])

print("\nFirst 10 actual labels:")
print(y_test.iloc[:10].to_numpy())
```

تولید پیش‌بینی و مقایسه اولیه

مدل پیش‌بینی‌های خود را روی داده آزمون تولید کرد. با مقایسه ۱۰ پیش‌بینی اول با برچسب‌های واقعی، می‌بینیم که مدل در این نمونه‌ها عملکرد کاملاً درستی داشته است. اما برای قضاوت نهایی باید معیارهای کلی را محاسبه کنیم.

بررسی Confusion Matrix و تحلیل خطا

ماتریس درهم‌ریختگی به شکل [[41, 1], [1, 71]] است. تفسیر آن به این صورت است:

۴۱ نمونه (True Negative): واقعاً malignant بودند و مدل هم malignant پیش‌بینی کرد.

۱ نمونه (False Positive): واقعاً benign بود اما مدل malignant پیش‌بینی کرد.

۱ نمونه (False Negative): واقعاً malignant بود اما مدل benign پیش‌بینی کرد.

۷۱ نمونه (True Positive): واقعاً benign بودند و مدل هم benign پیش‌بینی کرد.

مدل فقط ۲ خطا از ۱۱۴ نمونه دارد (حدود ۱.۷۵٪). این عملکرد فوق‌العاده است. نکته‌ی مهم این است که خطای False Negative (نادیده گرفتن بیمار واقعی) در کاربردهای پزشکی خطرناک‌تر است، اما در این مدل تعداد این خطا بسیار کم است.



```python
error_analysis = X_test.copy()
error_analysis["actual"] = y_test
error_analysis["predicted"] = predictions

# فقط ردیف‌هایی که actual با predicted برابر نیست
errors = error_analysis[error_analysis["actual"] != error_analysis["predicted"]]

print("Number of errors:", len(errors))
print("\nFirst 5 error samples:")
errors.head()
```

ر این بخش، نمونه‌هایی را که مدل اشتباه طبقه‌بندی کرده است، استخراج می‌کنیم. ابتدا یک کپی از X_test می‌سازیم و ستون‌های actual (برچسب واقعی) و predicted (پیش‌بینی مدل) را به آن اضافه می‌کنیم. سپس با شرط actual != predicted، فقط ردیف‌هایی را نگه می‌داریم که مدل در آن‌ها اشتباه کرده است. در نهایت، تعداد کل خطاها و ۵ نمونه‌ی اول را نمایش می‌دهیم تا بتوانیم بررسی کنیم مدل دقیقاً کجاها و چرا اشتباه می‌کند. این کار بخشی از تحلیل خطا (Error Analysis) است و به بهبود آینده‌ی مدل کمک می‌کند.



```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# ساخت Pipeline برای Cross-validation
cv_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000))
])

# تنظیم StratifiedKFold
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# اجرای Cross-validation فقط روی داده آموزش
scores = cross_val_score(
    cv_model,
    X_train,
    y_train,
    cv=cv,
    scoring="f1"
)

print("F1 scores:", scores)
print("Mean F1:", scores.mean())
print("Standard deviation:", scores.std())
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

# تنظیم StratifiedKFold
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# تعریف لیست معیارها
scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1'
}

# اجرای Cross-validation با چند معیار
results = cross_validate(
    cv_model,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    return_train_score=False  # فقط نتایج آزمون (اعتبارسنجی) را برگردان
)

print(results)
```
